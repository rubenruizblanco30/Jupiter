```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("dataset.csv")

df.head()
```

```python
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

df.info()
```

```python
df.describe()
```

```python
df = df.replace(r"^\s*$", pd.NA, regex=True)

print(df.isna().sum())
```

```python
print("Filas antes:", len(df))

columnas = df.columns.drop("row_id")
df = df.drop_duplicates(subset=columnas).copy()

print("Filas después:", len(df))
```

```python
df["city"] = df["city"].fillna("Desconocido")
df["subdivision"] = df["subdivision"].fillna("Desconocido")
df["submarket"] = df["submarket"].fillna("Desconocido")

df["sale_date"] = pd.to_datetime(df["sale_date"], errors="coerce")

df = df.dropna(subset=["sale_price", "sqft"]).copy()
df = df[(df["sale_price"] > 0) & (df["sqft"] > 0)].copy()

df["sale_year"] = df["sale_date"].dt.year

print("Filas tras la limpieza:", len(df))
```

```python
construccion_incorrecta = df["year_built"] > df["sale_year"]

reforma_incorrecta = (df["year_reno"] != 0) & (
    (df["year_reno"] > df["sale_year"]) |
    (df["year_reno"] < df["year_built"])
)

print("Construcciones posteriores a la venta:", construccion_incorrecta.sum())
print("Reformas incoherentes:", reforma_incorrecta.sum())

df.loc[construccion_incorrecta, "year_built"] = float("nan")
df.loc[reforma_incorrecta, "year_reno"] = float("nan")
```

```python
plt.hist(df["sale_price"], bins=50)

plt.title("Distribución del precio")
plt.xlabel("Precio")
plt.ylabel("Número de viviendas")

plt.show()
```

```python
plt.hist(df["sqft"], bins=50)

plt.title("Distribución de la superficie")
plt.xlabel("Superficie en pies cuadrados")
plt.ylabel("Número de viviendas")

plt.show()
```

```python
plt.scatter(df["sqft"], df["sale_price"], alpha=0.2)

plt.title("Superficie y precio")
plt.xlabel("Superficie en pies cuadrados")
plt.ylabel("Precio")

plt.show()
```
